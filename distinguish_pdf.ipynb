```python
import os
import fitz
from langdetect import detect
import shutil

source_folder = "/content/drive/MyDrive/disaster_data_nepal"
destination_folder = "/content/drive/MyDrive/english_disaster_manage_data"
os.makedirs(destination_folder, exist_ok=True)


def lang_classifier():

  for pdf in os.listdir(source_folder):
    if pdf.lower().endswith('.pdf'):
      full_pdf_path = os.path.join(source_folder, pdf)
      try:
        doc = fitz.open(full_pdf_path)
        text = ""
        for page in doc:
          text += page.get_text()
        doc.close()
        if text.strip():
          language = detect(text)
          if language == "en":
            shutil.move(full_pdf_path, destination_folder)
            print(f"Moved English PDF: {pdf}")
      except Exception as e:
        print(f"Error processing {pdf} : {e}")

lang_classifier()
```
